## 3. Verify Dataset Configuration (`data.yaml`)

```python
import yaml

with open('data.yaml', 'r') as f:
    data_config = yaml.safe_load(f)

# Update path to current directory on Kaggle
data_config['path'] = '.'
with open('data.yaml', 'w') as f:
    yaml.dump(data_config, f)

print(data_config)
```

## 4. Train the YOLOv8 Segmentation Model

```python
from ultralytics import YOLO

# Load pretrained YOLOv8 nano segmentation model
model = YOLO('yolov8n-seg.pt')

# Train model
results = model.train(
    data='data.yaml',
    epochs=50,
    imgsz=640,
    batch=16,
    plots=True
)
```

## 5. Visualize Validation Predictions & Results

```python
from IPython.display import Image, display
import glob

# Show prediction batch
pred_images = glob.glob('runs/segment/**/val_batch0_pred.jpg', recursive=True)
if pred_images:
    display(Image(filename=pred_images[-1]))

# Show training results graphs
result_plots = glob.glob('runs/segment/**/results.png', recursive=True)
if result_plots:
    display(Image(filename=result_plots[-1]))
```

## 6. Run Inference on Test Images

```python
# Load the best trained weights
best_model_weights = glob.glob('runs/segment/**/weights/best.pt', recursive=True)
model = YOLO(best_model_weights[-1] if best_model_weights else 'yolov8n-seg.pt')

# Predict on a sample image
preds = model.predict(source='images/train/RedDanecattle100.jpg', save=True, conf=0.5)

# Display prediction
display(Image(filename=str(preds[0].save_dir / 'RedDanecattle100.jpg')))
```
